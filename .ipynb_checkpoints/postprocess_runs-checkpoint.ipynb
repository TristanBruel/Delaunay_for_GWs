```python
burn = 1_000
nsteps = 3_000
all_triangulations = []
all_corner_weights = []
all_beta_q = []
all_chi = []
all_tilt = []

for run_label in range(num_runs):
    with open(f"results_longer_run_{run_label}/backend", "rb") as f:
        back = pickle.load(f)
        nsteps = back.inds["tri"].shape[0]
        
        triangulations = [
            back.chain["tri"][step, 0, walker][back.inds["tri"][step, 0, walker]]
            for step, walker in product(range(burn, nsteps), range(nwalkers))
        ]
        
        corner_weights = np.array([
            back.chain["corner_w"][step, 0, walker]
            for step, walker in product(range(burn, nsteps), range(nwalkers))
        ])
        
        beta_q = np.array([
            back.chain["beta_q"][step, 0, walker]
            for step, walker in product(range(burn, nsteps), range(nwalkers))
        ])
        
        chi = np.array([
            back.chain["chi"][step, 0, walker]
            for step, walker in product(range(burn, nsteps), range(nwalkers))
        ])
        
        tilt = np.array([
            back.chain["tilt"][step, 0, walker]
            for step, walker in product(range(burn, nsteps), range(nwalkers))
        ])
        
        all_triangulations.extend([
            np.vstack([
                triangulations[ind], 
                np.c_[corners, corner_weights[ind].T]
            ])
            for ind in range(len(triangulations))
        ])
        all_corner_weights.append(corner_weights)
        all_beta_q.append(beta_q)
        all_chi.append(chi)
        all_tilt.append(tilt)

# Concatenate all runs into single arrays
triangulations = all_triangulations
corner_weights = np.concatenate(all_corner_weights, axis=0)
beta_q = np.concatenate(all_beta_q, axis=0)
chi = np.concatenate(all_chi, axis=0)
tilt = np.concatenate(all_tilt, axis=0)
```

```python
def q_log_prior(q, m1, beta_q, m_min=m1_min):
    out_bounds = q * m1 < m_min

    normalisation = (beta_q + 1) / (1 - (m_min / m1) ** (beta_q + 1))
    p_of_q = normalisation * q**beta_q

    return np.where(out_bounds, -1e300, np.log(p_of_q))
```

```python
q = np.linspace(0, 1, 200)

m1s = [10, 30, 40]
quantiles = np.zeros((len(m1s), 3, len(q)))

for ind, m1 in enumerate(m1s):
    q_log_pdf = np.array([
        q_log_prior(q, m1, b[0, 0])
        for b in beta_q
    ])
    
    quantiles[ind] = np.quantile(np.exp(q_log_pdf), [0.05, 0.5, 0.95], axis=0)    
    
np.savez("quantiles_q", quantiles=quantiles, q=q, m1s=m1s)
```

```python
z_grid = np.linspace(0.1, z_max, 50)
m1_grid = np.linspace(m1_min, m1_max, 100)

M1, Z = np.meshgrid(m1_grid, z_grid)
grid = np.c_[M1.ravel(), Z.ravel()]
```

```python
from astropy.cosmology import Planck15
from astropy import units as u

# >>> file_vt=h5py.File('o1+o2+o3_bbhpop_real+semianalytic-LIGO-T2100377-v2.hdf5')
# >>> Tobs=file_vt['injections'].attrs["analysis_time_s"]
Tobs = 66997600.0 / 86400 / 365 #63129577.0 / 86400 / 365 # years
to_comoving_volume = (1 + z_grid) / (
    4 * np.pi * Tobs * Planck15.differential_comoving_volume(z_grid).to(u.Gpc**3 / u.sr).value
)
```

```python
def compute_num_events(triangulation_points):
    tri = delaunaytor.CPUDelaunayInterpolator()
    tri.triangulate(triangulation_points)
    weights_in_vertices = tri.weights[tri.triangulation.simplices]

    weight_diffs = np.c_[
        (weights_in_vertices[:, 1] - weights_in_vertices[:, 2]),
        (weights_in_vertices[:, 2] - weights_in_vertices[:, 0]),
        (weights_in_vertices[:, 0] - weights_in_vertices[:, 1]),
    ]
    bar_integral = (np.exp(weights_in_vertices) * weight_diffs).sum(axis=-1) / (
        -weight_diffs
    ).prod(axis=-1)

    return (2 * tri.volumes() * bar_integral).sum()
```

```python
num_triangulations = 10_000
selected_tris = np.random.choice(num_triangulations, size=num_triangulations, replace=False)

log10_rate = np.zeros((num_triangulations, grid.shape[0]))


for ind, tri_ind in enumerate(tqdm(selected_tris)):
    this_delo = delaunaytor.CPUDelaunayInterpolator()
    this_delo.triangulate(triangulations[tri_ind])
    log10_rate[ind] = (this_delo.interpolate(grid)) / np.log(10)
```
